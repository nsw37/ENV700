```python
# 1.1 Fetch and parse the website
url = "https://www.eia.gov/electricity/state/"

response = requests.get(url)
response.raise_for_status()

webpage = BeautifulSoup(response.text, "html.parser")
type(webpage)

if webpage.title: 
    print(webpage.title.get_text(strip=True))
else: print("No page title")

# 1.2 Locate elements and read their text attributes into variables
state_names_nodes = webpage.select(
    ".basic-table a")
state_name = [state.get_text(strip=True) for state in state_names_nodes]
state_name

avg_retail_node = webpage.select(
    "td:nth-child(2)")
avg_retail_price = [retail.get_text(strip=True) for retail in avg_retail_node]
avg_retail_price

net_summary_node = webpage.select(
    "td:nth-child(3)")
net_summary_capacity = [capacity.get_text(strip=True) for capacity in net_summary_node]
net_summary_capacity

net_generation_node = webpage.select(
    "td:nth-child(4)")
net_generation = [generation.get_text(strip=True) for generation in net_generation_node]
net_generation

total_retail_node = webpage.select(
    "td:nth-child(4)")
total_retail = [total.get_text(strip=True) for total in net_summary_node]
total_retail
# 1.3 Construct a dataframe from the values
electricity_usage = pd.DataFrame({
    "State" : state_name,
    "Average Retail Price" : pd.to_numeric(avg_retail_price),
    "Net Summary Capacity" : pd.to_numeric(net_summary_capacity),
    "Net Generation" : pd.to_numeric(net_generation),
    "Total Retail Sales" : pd.to_numeric(total_retail)
})

# 1.4 Wrangle


# 1.5 Examine


# 1.6. Plot
```
